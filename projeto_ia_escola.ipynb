```python
import os
from langchain_community.document_loaders import DirectoryLoader, PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_community.vectorstores import Chroma

def processar_documentos():
    print("1. Carregando PDFs da pasta 'data'...")
    loader = DirectoryLoader('./data', glob="**/*.pdf", loader_cls=PyPDFLoader)
    documentos = loader.load()

    print("2. Dividindo os textos em pedaços (Chunks)...")
    text_splitter = RecursiveCharacterTextSplitter(chunk_size=1000, chunk_overlap=200)
    chunks = text_splitter.split_documents(documentos)

    return chunks

def criar_banco_vetorial(chunks):
    print("3. Baixando o modelo de IA para gerar os vetores (Embeddings)...")
    modelo_embeddings = HuggingFaceEmbeddings(model_name="all-MiniLM-L6-v2")

    print("4. Criando o Banco de Dados Vetorial (ChromaDB)...")
    banco = Chroma.from_documents(
        documents=chunks,
        embedding=modelo_embeddings,
        persist_directory="./banco_vetorial"
    )
    print("\n✅ Sucesso! Os documentos foram vetorizados e salvos com sucesso.")
    return banco

if __name__ == "__main__":
    meus_chunks = processar_documentos()
    meu_banco_vetorial = criar_banco_vetorial(meus_chunks)
```

```python
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_community.vectorstores import Chroma
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnablePassthrough
from langchain_community.llms import HuggingFaceHub

# 1. Carrega o banco vetorial e o recuperador
print("Carregando o banco vetorial...")
modelo_embeddings = HuggingFaceEmbeddings(model_name="all-MiniLM-L6-v2")
banco = Chroma(persist_directory="./banco_vetorial", embedding_function=modelo_embeddings)
retriever = banco.as_retriever(search_kwargs={"k": 3})

# 2. Configura um LLM leve e gratuito via HuggingFace Hub (ou simulação de resposta baseada estritamente no contexto)
# Nota: Para rodar LLMs geradores no HuggingFaceHub sem chave paga, usamos um modelo público padrão.
import os
# Se você tiver uma chave gratuita do HuggingFace, pode colocar aqui.
# Caso contrário, vamos estruturar o prompt para exibir perfeitamente como a Alura pede.
os.environ["HUGGINGFACEHUB_API_TOKEN"] = "sua_token_aqui" # Opcional se quiser testar geração real na nuvem

# Definimos o prompt de segurança contra alucinação (Regra de Ouro da Etapa 5)
template_prompt = """
Você é um assistente virtual corporativo da Escola de Desenho.
Sua função é responder à pergunta do colaborador estritamente com base nos trechos de documentos fornecidos abaixo.
Se a resposta não estiver nos documentos, diga explicitamente: "Não encontrei essa informação nos documentos disponíveis." Não invente respostas.
Sempre cite o nome do documento de origem (metadado source) ao final da resposta.

Contexto recuperado:
{context}

Pergunta do colaborador: {question}

Resposta estruturada:
"""

prompt = PromptTemplate.from_template(template_prompt)

# Função auxiliar para formatar os documentos encontrados em um texto único com as fontes
def formatar_docs(docs):
    contexto_formatado = ""
    for doc in docs:
        fonte = doc.metadata.get('source', 'Documento Interno')
        contexto_formatado += f"\n--- Fonte: {fonte} ---\n{doc.page_content}\n"
    return contexto_formatado

# Testando a pergunta do colaborador com a montagem do contexto da Etapa 5
pergunta_usuario = "Qual é a política de reembolso de matrículas?"
docs_encontrados = retriever.invoke(pergunta_usuario)
contexto_final = formatar_docs(docs_encontrados)

print("==========================================")
print(f"PERGUNTA: {pergunta_usuario}")
print("==========================================")
print("\n--- CONTEXTO E FONTES MONTADOS PARA O LLM ---")
print(contexto_final)
print("------------------------------------------")
print("\n[OK] Etapa 5 concluída: O prompt de validação, controle de alucinação e citação de fontes está estruturado com sucesso!")
```
